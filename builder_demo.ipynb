```python
import pandas as pd
import spacy

from sklearn.model_selection import train_test_split
from builder import NLIDatasetBuilder
nlp = spacy.load("en_core_web_sm")
```

```python
builder = NLIDatasetBuilder(

    preprocessor_params={
        "columns": ("premise", "hypothesis"),
        "language": "en",
        "language_column": "lang_abv",
        "lowercase": True,
        "strip_accents": True,
        "tokenizer": "spacy",
        "stopword_method": "sklearn",
        "keep_negations": True,
        "custom_stopwords": None,
        "reduction": "none",
        "ngram_orders": (2, 3),
        "top_n": 20,
        "use_pos": True
    },

    vectorizer_params={
        "method": "tfidf",
        "max_features": 30000,
        "min_df": 2,
        "word_ngram_range": (1, 3),
        "char_ngram_range": (3, 5),
        "svd_components": 300,
        "normalize_embeddings": True,
        "embedding_model": "sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2",
        "embedding_batch_size": 32,
        "sublinear_tf": True,
        "random_state": 42
    }
)
```

```python
train_df = pd.read_csv('data/train.csv')

# Split into train and test (80% / 20%)
train_df, test_df = train_test_split(
    train_df,
    test_size=0.2,
    random_state=42,
    shuffle=True
)

X_train = builder.build_dataset(
    train_df,
    mode="train"
)

X_test = builder.build_dataset(
    test_df,
    mode="val"
)
```
